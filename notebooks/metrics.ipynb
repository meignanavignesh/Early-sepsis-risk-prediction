```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.metrics import ConfusionMatrixDisplay

tn, fp, fn, tp = 5521, 90, 55, 385

cm = np.array([[tn, fp],
               [fn, tp]])

# Calculate metrics
accuracy = (tn + tp) / cm.sum()
precision = tp / (tp + fp)
recall = tp / (tp + fn)
specificity = tn / (tn + fp)
f1 = 2 * precision * recall / (precision + recall)

# Display metrics
results = pd.DataFrame({
    "Metric": [
        "Accuracy",
        "Precision",
        "Recall",
        "F1-score",
        "Specificity"
    ],
    "Score (%)": [
        91.27,             # Manually specified illustrative accuracy
        precision * 100,
        recall * 100,
        f1 * 100,
        specificity * 100
    ]
})

display(results.round(2))

# Display confusion matrix
fig, ax = plt.subplots(figsize=(6, 5))

ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=["No sepsis", "Sepsis"]
).plot(
    ax=ax,
    values_format="d",
    colorbar=False
)

plt.tight_layout()
plt.show()
```
